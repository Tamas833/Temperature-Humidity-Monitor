```python
import random
import sqlite3
import time
import math

import pandas as pd
import matplotlib.pyplot as plt

from datetime import datetime
from zoneinfo import ZoneInfo
from IPython.display import display, clear_output


ADATBAZIS_FAJL="/content/monitor_szimulacio.db"

MERESEK_SZAMA=30

MERESI_IDO=1


def adatbazis_letrehozasa():
  """Letrehozza a mereseket tarolo tablat."""

  with sqlite3.connect(ADATBAZIS_FAJL) as kapcsolat:
    kapcsolat.execute("""
    CREATE TABLE IF NOT EXISTS meresek(
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    idopont TEXT NOT NULL,
    homerseklet REAL NOT NULL,
    paratartalom REAL NOT NULL
    CHECK (paratartalom BETWEEN 0 AND 100))""")

def random_meres():
  """Veletlen tesztadatokat keszit."""

  homerseklet=round(random.uniform(-20,40),1)

  paratartalom=round(random.uniform(20,90),1)

  return homerseklet,paratartalom

def adat_mentese(idopont,homerseklet,paratartalom):
  """Elment egy adatpart az SQLITE-adatbazisba."""

  with sqlite3.connect(ADATBAZIS_FAJL) as kapcsolat:
    kapcsolat.execute(
        """
        INSERT INTO meresek (
        idopont,
        homerseklet,
        paratartalom
        )
        VALUES (?,?,?)
        """,
        (idopont,homerseklet,paratartalom)
    )

def diagram_rajzolasa(idopontok,homersekletek,paratartalmak):
  """Kirajzolja az aktualis futas ket diagramjat."""

  abra,tengelyek=plt.subplots(2,1,figsize=(10,7),sharex=True)

  tengelyek[0].plot(idopontok,homersekletek,color="red",marker="o")

  tengelyek[0].set_title("Homerseklet alakulasa")

  tengelyek[0].set_ylabel("Homerseklet ('C)")
  tengelyek[0].set_ylim(-25,45)
  tengelyek[0].axhline(0,color="gray",linestyle="--")
  tengelyek[0].grid(True)

  tengelyek[1].plot(idopontok,paratartalmak,color="blue",marker="o")


  tengelyek[1].set_title("Relativ paratartalom kialakulasa")

  tengelyek[1].set_ylabel("Paratartalom (%)")

  tengelyek[1].set_xlabel("Idopont")
  tengelyek[1].set_ylim(0,100)
  tengelyek[1].grid(True)

  import matplotlib.dates as mdates


  tengelyek[1].xaxis.set_major_formatter(
      mdates.DateFormatter(
          "%H:%M:%S",
          tz=ZoneInfo("Europe/Bucharest")
      )
  )

  abra.suptitle("SZIMULACIO-veletlen tesztadatok")
  abra.autofmt_xdate()
  abra.tight_layout()

  display(abra)
  plt.close(abra)



def mentett_adatok_megjelenitese():
  """Megmutatja az adatbazis utolso 10 sorat."""

  with sqlite3.connect(ADATBAZIS_FAJL) as kapcsolat:
    tabla=pd.read_sql_querry(
        """
        SELECT
        id AS "Azonosito",
        idopont AS "Idopont",
        homerseklet AS "Homerseklet ('C)",
        paratartalom AS "Paratartalom (%)"
        FROM meresek
        ORDER BY id DESC
        LIMIT 10
        """,
        kapcsolat
    )

    display(tabla)


def main():
  adatbazis_letrehozasa()

  idopontok=[]
  homersekletek=[]
  paratartalmak=[]

  try:
      for sorszam in range(1,MERESEK_SZAMA+1):
            idopont=datetime.now(
            ZoneInfo("Europe/Bucharest")
      )

            homerseklet,paratartalom=random_meres()

            adat_mentese(
          idopont.isoformat(timespec="milliseconds"),homerseklet,paratartalom
      )

            idopontok.append(idopont)

            homersekletek.append(homerseklet)

            paratartalmak.append(paratartalom)

            clear_output(wait=True)

            print("HOMERSEKLET- ES PARATARTALOM-MONITOR")
            print("SZIMULACIO- nem valodi erzekeloadatok")
            print(f"\nMeres: {sorszam}/{MERESEK_SZAMA}")
            print(f"Homerseklet: {homerseklet} 'C")

            print(f"Paratartalom: {paratartalom} %")

            print("Az adatpart elmentettuk az adatbazisba.")

            diagram_rajzolasa(
          idopontok,
          homersekletek,
          paratartalmak
      )

            if sorszam<MERESEK_SZAMA:

              time.sleep(MERESI_IDO)

  except KeyboardInterrupt:
          print("\nA szimulaciot leallitottad.")

  except sqlite3.Error as hiba:

          print(f"\nAdatbazishiba: {hiba}")

  return


  print("\nA szimulacio befejezodott.")
  print("Az utolso 10 mentett adat:")


  mentett_adatok_megjelenitese()

main()





```
